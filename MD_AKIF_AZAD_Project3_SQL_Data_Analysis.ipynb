```python
# Step 1: Tools Ko Bulao ...
import pandas as pd
import sqlite3
# Step 2: DATA Laao ...
df = pd.read_excel('Dataset for Data Analytics.xlsx')
# Step 3: SQLITE3 Aurr PYTHON Dono Jodna Yaani New Kitchen Bana Na...
conn = sqlite3.connect('decode_database.db')
# Step 4: DATA Ko Excel Se SQL Mein Daalna ...
df.to_sql('ecommerce_table', conn, if_exists='replace', index=False)
# Step 5: DATA Ko Filter Karna Jo Hamare Kaam Ka Hai ...
query = """ SELECT Product, SUM(TotalPrice) AS TotalSales
FROM ecommerce_table
WHERE Orderstatus = 'Delivered'
GROUP BY Product
ORDER BY TotalSales DESC; """
# Step 6:FINAL Result Hidden Insight Dekhna ...
result = pd.read_sql_query(query,conn)
display (result)
```
