```python
import pandas as pd
import numpy as np
df = pd.read_excel('Dataset for Data Analytics.xlsx')
print('\n--- DUMMY DATA ---\n')
display(df)
print('\n--- DATA Ka Information ---\n')
df.info()
df.drop_duplicates()
print('\n--- Missing DATA Dhund Kar Dikhao ---\n')
print(df.isnull().sum())
df ['CouponCode'] = df['CouponCode'].fillna('No Coupon')
df.reset_index()
print('\n--- CLEAN DATA ---\n')
display(df)
print('\n--- CLEAN DATA Ka Information Dikhao ---\n')
df.info()
print('\n--- CLEAN DATA Pe EDA Karna ---\n')
print(df.describe())
print(df.corr(numeric_only=True))
print('\n--- OUTLIER DETECTION (IQR ENGINE) ---\n')
Q1 = df['TotalPrice'].quantile(0.25)
Q3 = df['TotalPrice'].quantile(0.75)
IQR = Q3 - Q1
upper_limit = Q3 + (1.5 * IQR)
outliers = df[df['TotalPrice'] > upper_limit]
print(outliers)
```

📊 Project 2: Exploratory Data Analysis (EDA) Findings
1. 🧮 Basic Statistics & Data Shape:
After running df.describe(), the numerical columns (Quantity, UnitPrice, ItemsInCart, TotalPrice) were analyzed.
The TotalPrice column shows a significant gap between the Mean (1053.96) and the Median (823.61). This indicates a right-skewed distribution, meaning a few large orders are pulling the average up.
2. 🚨 Outlier Detection (Using IQR Method):
Because TotalPrice is skewed, the IQR (Interquartile Range) method was applied to find anomalies.
The upper bound (safe limit) was mathematically calculated at 3330.40.
Finding: The maximum Total Price in the dataset is 3456.40. The IQR engine successfully flagged 8 transactions as Outliers because they exceeded the upper limit. These represent VIP customers or bulk orders and should not be deleted.
3. 🤝 Correlation Analysis:
Using df.corr(), the relationships between numerical variables were tested.
Strong Positive Correlation: There is a high correlation of 0.71 between UnitPrice and TotalPrice. This mathematically confirms that higher-priced individual items strongly drive up the final bill.
Logical Correlation: Quantity and ItemsInCart also show a strong correlation (0.65), indicating customers buying more quantity tend to have more unique items in their cart.
No Correlation: UnitPrice and ItemsInCart have a near-zero score (0.0006), proving that buying expensive items does not mean the customer will buy a larger variety of items.
